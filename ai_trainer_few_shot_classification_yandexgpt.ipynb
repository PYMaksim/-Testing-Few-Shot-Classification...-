```python

"""
Project: LLM Comparison (Few-Shot Classification)
"""

from google.colab import userdata
from openai import OpenAI
import time

# --- CONFIGURATION ---
API_KEY = userdata.get('Y')
FOLDER_ID = "b1g2hgrbnnrqll94nr9k"
BASE_URL = "https://ai.api.cloud.yandex.net/v1"

HEADERS = {
    "Authorization": f"Api-Key {API_KEY}",
    "x-folder-id": FOLDER_ID
}

client = OpenAI(api_key=API_KEY, base_url=BASE_URL)
MODEL_ID = "gpt://b1g2hgrbnnrqll94nr9k/yandexgpt-5-pro/latest"
```

```python
# --- EXAMPLES ---
examples = [
    {"role": "user", "content": "Деньги списали, но заказ не оформился."},
    {"role": "assistant", "content": "Категория: Возврат средств"},
    {"role": "user", "content": "Когда приедет курьер? Номер 12345."},
    {"role": "assistant", "content": "Категория: Статус доставки"},
    {"role": "user", "content": "Приложение вылетает при оплате."},
    {"role": "assistant", "content": "Категория: Техническая ошибка"}
]

test_question = {"role": "user", "content": "Хочу отменить заказ №987, я только что его оплатил."}
```

```python
# --- EXECUTION ---
print("🚀 Testing Few-Shot Classification...")

messages = examples + [test_question]

response = client.chat.completions.create(
    model=MODEL_ID,
    messages=messages,
    temperature=0.0,
    max_tokens=20
)

print("✅ Результат:")
print(response.choices[0].message.content)
```
